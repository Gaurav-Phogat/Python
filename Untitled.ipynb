## Testing class and objects in Python




```python
class car:
    max_speed = 120

    def __init__(self,model,color,car_type,speed = 0):
        if(speed > car.max_speed):
            self.speed = car.max_speed
        else: 
            self.speed = speed
            
        self.color = color
        self.model = model
        self.car_type = car_type

    def accelerate(self,acceleration):
        if(self.speed + acceleration <= car.max_speed):
            self.speed += acceleration
        else:
            self.speed = car.max_speed

    def changeProperties(self,model,color,car_type,speed):
        self.model = model
        self.color = color
        self.car_type = car_type

        if(speed > car.max_speed):
            self.speed = car.max_speed
        else: 
            self.speed = speed

    def getData(self):
        return [self.model,self.color,self.car_type,self.speed]
```

```python
firstCar = car("Tata zest","silver","sedan",speed = 50)
Data = firstCar.getData()
print(Data)
```
