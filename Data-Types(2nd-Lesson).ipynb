## Data Types (2nd lesson)

Below are the data types and typecastings.

number types: int, float.

word types: String.

boolean type: true or false.

```python
# typecasting from float to int and int to float
num1 = 1.1
print(int(num1))

num2 = 2
print(float(num2))
```

```python
#typecasting from int to string and float to string but type casting string to int or float causes error if the string did not have a number value.

num1 = 1.12

print(str(num1))

str1 = "2.25"

print(float(str1))

str2 = "A"

print(float(str2))
```

```python
#typecasting boolean to str or int.

bool1 = True

print(int(bool1))

bool2 = False

print(int(bool2))

# to string

print(str(bool1))

# test area ignore.

type(bool1)
```
